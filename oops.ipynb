```python
# OOPs (Object-Oriented Programming) programming ka ek approach hai jisme program ko objects 
# aur classes ke around design kiya jata hai.

# OOPs ke 4 main pillars:
# Encapsulation → Data ko protect/bind karna.
# Inheritance → Parent class ki properties/methods child class mein use karna.
# Polymorphism → Same method/interface ka different behavior.
# Abstraction → Unnecessary details ko hide karke important cheez dikhana.
```

```python
# No.	[Class]	                                                             [Object]
# 1	Class ek blueprint/design hoti hai              	        Object class ka actual instance hota hai
# 2	Class logical structure hoti hai	                        Object real entity hoti hai
# 3	Class mein attributes aur methods define hote hain	        Object unhe access/use karta hai
# 4	Class se multiple objects ban sakte hain	                Har object class ka separate instance hota hai
# 5	Class example: Student                                   	Object example: s1 = Student()
# 6	Class batati hai kya banana hai	                            Object batata hai actual bana hua kya hai
```

```python
# [class]                                                         [ object]
# Blueprint/design hoti hai	                               Class ka instance hota hai
# Structure define karti hai	                               Structure ko actually use karta hai
# Attributes aur methods define hote hain	                   Attributes aur methods access karta hai
# Ek class se multiple objects ban sakte hain	Har                object alag instance hota hai
# Example: Student	                                         Example: s1 = Student()
# Class = design	                                             Object = real entity
```

```python
# 2. [Method vs]                                                         [Function]
# Method                                                             	Function
# Class ke andar define hota hai	                                    Generally class ke bahar define hota hai
# Object/class se related hota hai                                    General-purpose hota hai
# self ya cls use kar sakta hai	                                    Normally self/cls nahi hota
# Object ke data par kaam kar sakta hai	                             Independent task perform karta hai
# obj.show() se call ho sakta hai	                                     show() se call hota hai
# Method class ka member hota hai	                                      Function independent hota hai
```

```python
# 4.[ Encapsulation vs]                                                         Abstraction
# Encapsulation                                                        [Abstraction]
# Data aur methods ko ek unit/class mein combine karta hai	               Unnecessary implementation details hide karta hai
# Data protection par focus	                                              Essential features dikhane par focus
# Private/protected members use ho sakte hain	                             Abstract class/method use ho sakte hain
# Example: __balance	                                                     Example: abstract withdraw()
# Question: Data ko protect kaise karein?                              	Question: User ko kya dikhana hai?
# Main idea = Wrapping + Protection                                    	Main idea = Hiding complexity
```

```python
# 5. [Inheritance vs ]                                                 [Polymorphism]
# Inheritance                                                  	[Polymorphism]
# Parent class se properties/methods lena	                  Same interface/name ka different behavior
# Code reuse mein help karta hai                                  	Flexibility provide karta hai
# Parent-child relationship hota hai	                            One name, multiple forms
# class Child(Parent)	                             Same method different classes mein different behavior
# Existing code ko reuse karta hai	           Existing interface ko different tarike se implement kar sakta hai
# Main idea = Reuse                              Main idea = Many forms
```

```python
# 6. Instance Method vs                                                             Class Method
# Instance Method                                                               	[Class Method]
# self parameter use karta hai	                                        cls parameter use karta hai
# Object ke data par kaam karta hai                                 	Class ke data par kaam karta hai
# @classmethod ki zarurat nahi	                                   @classmethod decorator lagta hai
# Usually object se call hota hai	                                Class ya object dono se call ho sakta hai
# Har object ke instance data ko access kar sakta hai	            Class variable ko access/change kar sakta hai
# Example: obj.show()	                                               Example: Student.show()
```

```python
# nstance Method vs                                                          Static Method
# Instance Method                                                 	[Static Method]
# 1. self parameter use karta hai                                    	self ya cls parameter use nahi karta
# 2. Object ke data ko access kar sakta hai	                         Directly instance data access nahi karta
# 3. @staticmethod decorator nahi lagta	                             @staticmethod decorator lagta hai
# 4. Usually object ke through call hota hai	                Class ya object dono ke through call ho sakta hai
# 5. Instance variables aur methods access kar sakta hai	       Mainly independent/general task perform karta hai
# 6. Example: obj.show()                                        	Example: Student.show()
```

```python
# ⭐ Sabse easy revision

#Class → Design
# Object → Real instance

# Function → General
# Method → Class-related

# Constructor → Start
# Destructor → Cleanup/End

# Encapsulation → Data protection
# Abstraction → Complexity hiding

# Inheritance → Reuse
# Polymorphism → Many forms

# Instance Method → self
# Class Method → cls

# Overloading → Different inputs
# Overriding → Child changes parent method
```

```python
from abc import ABC, abstractmethod

class bankaccount(ABC):
    def __init__(self,account_holder, balance):
        self.account_hplder=account_holder
        self.__balance=balance


    def deposit(self,amount):
        self.__balance+=amount
        print("deposit:{amount}")

    def get_balance(self):
        return self.__balance

    @abstractmethod
    def calculate_interst(self):
        pass

    # INHERITANCE

class savingaccount(bankaccount):
        def calculate_interst(self):
            return self.get_balance() *0.05


class currentaccount(bankaccount):
     def calculate_interst(self):
          return self.get_balance()  *0.02

# POLYMORPHISM

accounts=[
    savingaccount("priyanka",50000),
    currentaccount("rahul",50000)
]

for account in accounts:
     print("Account holder:",account.account_holder)
     print("balance:",account.get_balance())
     print("interst:",account.calculate_interst())
     print()



    
```
